# 01 - Create Catalog Structure
## SalesFlow Data Lakehouse | Setup Phase

This notebook creates the foundational Unity Catalog structure for the SalesFlow Lakehouse project,
migrating sales data from SQL Server on-premise to Databricks.

### What this notebook does:
- Creates the `salesflow_dev` catalog
- Creates schemas: `bronze`, `silver`, and `gold` with descriptive comments
- Creates the `sales_data` volume inside the `bronze` schema
- Validates the structure by listing schemas and volumes created

### Lakehouse Architecture:
| Schema | Layer | Purpose |
|--------|-------|---------|
| `bronze` | Raw | Data ingested as-is from source CSV files |
| `silver` | Curated | Cleaned, validated and enriched data |
| `gold` | Analytical | Dimensional model ready for analytics |

### Notes:
- Platform: Databricks Community Edition
- Catalog: `salesflow_dev`
- Source data: 8 CSV files extracted from SQL Server (semicolon-delimited)

### Output:
- Catalog and schemas visible in the Data Explorer
- Volume `sales_data` accessible at `/Volumes/salesflow_dev/bronze/sales_data/`

```sql
CREATE CATALOG IF NOT EXISTS salesflow_dev
COMMENT 'Development catalog for the SalesFlow Lakehouse project. Contains Bronze, Silver, and Gold schemas for the sales data migration from SQL Server on-premise.';
```

```sql
CREATE SCHEMA IF NOT EXISTS salesflow_dev.bronze
COMMENT 'Raw ingestion layer. Stores data as-is from SQL Server on-premise CSV extracts, with added control metadata (ingestion timestamp, source system, file name). No transformations applied.';
```

```sql
CREATE SCHEMA IF NOT EXISTS salesflow_dev.silver
COMMENT 'Curated layer. Stores cleaned, validated and standardized data from Bronze. Applies quality flags, string normalization, and phone standardization. Source of truth for business logic.';
```

```sql
CREATE SCHEMA IF NOT EXISTS salesflow_dev.gold
COMMENT 'Analytical layer. Stores the dimensional model (facts and dimensions) built from Silver. Optimized for BI consumption and analytics with surrogate keys and pre-aggregated structures.';
```

```sql
CREATE VOLUME salesflow_dev.bronze.sales_data
COMMENT 'Landing zone for the 8 source CSV files extracted from SQL Server on-premise (semicolon-delimited). Accessible at /Volumes/salesflow_dev/bronze/sales_data/.';
```
